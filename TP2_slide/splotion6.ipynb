```python
#step1
grades={
    "Dara" :[80 , 75,90],
    "Sophea":[95,88,92],
    "Vibols":[40 , 55 , 45],
    "Malis:":[70,85,78],
}
```

```python
#step2
averages={}
for name , score in grades.items():
    averages[name]=round(sum(score)/len(score),1)
print("Average scores of students: ", averages)
```

```python
#step3
passed =set()
for name , avg in averages.items():
    if avg>=50:
        print(name , " PASS")
        passed.add(name)
        
    else:
        print(name , " FAIL")
print("Passed :" , sorted(passed))
```

```python
#Step4
ranking =sorted(averages.items() , key =lambda item: item[1], reverse=True)
for rank , (name , avg) in enumerate(ranking , start=1):
    print(f"#{rank} {name} {avg}")
```
